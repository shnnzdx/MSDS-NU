## `TRY AND EXCEPT`
- try:<br>
    Put code here that could throw error(s)
    
- except Exception as e:<br>
   Put code here to handle exception(s) appropriatly - have specific error on the top and generic "Exception" at the end

- else: <br>
    Code to execute if code under try block is successful
    
    
- finally: <br>
    Code to execute whether code under try block is successful or fails

```python
# Lets make thigs more real (complicated!)
filepath = 'UsedCarInventory.txt'
list1=["A", "b", "c", "d"]

try:
    with open(filepath, 'r') as filehandle:
        for line in filehandle:
            #print(line) 
            pass
        print(list1[2])
        i = int("rt")
except FileNotFoundError as e:
    print("An error occured while opening the file", e)

except (IndexError, NameError) as e:
    print("Non generic error: ", e)

except Exception as e:
   print("Catch All error: ", e)

else: # Perform tasks here if code under try block is successful
    print("TO DO - Everything succeeded under try")
    
finally: # Perform tasks here whether code under try block is successful or fails
    print("TO DO - Execute under finally")
```

### `Transforming Data Using a Function or Mapping`

```python
#mapping-------------------------

data = pd.DataFrame({'food': ['bacon', 'pulled pork', 'bacon',
                              'Pastrami', 'corned beef', 'Bacon',
                              'pastrami', 'honey ham', 'nova lox'],
                     'ounces': [4, 3, 12, 6, 7.5, 8, 3, 5, 6]})
data

# 统一大小写
lowercased = data['food'].str.lower()   
data['food'] = lowercased

# 新增列by mapping using dictionary
meat_to_animal = {
  'bacon': 'pig',
  'pulled pork': 'pig',
  'pastrami': 'cow',
  'corned beef': 'cow',#
  'honey ham': 'pig',
  'nova lox': 'salmon' 
}  #map meat to animal
data['animal'] = lowercased.map(meat_to_animal) #map values from dictionary to new column 
```

```python
# function--------------------------------
def ounce_to_pounds(wt_oz):
    wt_lbs = wt_oz * 0.0625
    return wt_lbs

data['pounds'] = data['ounces'].map(ounce_to_pounds)
```
